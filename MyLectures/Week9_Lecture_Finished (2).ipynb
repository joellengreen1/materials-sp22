# Conditional Statements
Multi-line statement that allows Python to choose among different alternatives based on the truth value of an expression.  The format of a conditional statement is as follows.
```
if x > 0:
   statements
elif x < 0:
   statements
else:
   statements

```

````python
# create if statement to evaluate if value negative, zero, or positive then print out the results
x = 0
if x > 0:
    y = "positive"
elif x < 0:
    y = "negative"
else:
    y = "zero"

print("The number is",y)
````

````python
# create a function to pass a value and uses the if statement above
def sign(x):
    """assign positive, negative, and zero to value entered"""
    if x > 0:
        y = "positive"
    elif x < 0:
        y = "negative"
    else:
        y = "zero"  
    return y
    
````

````python
# run the function
sign(345)
````

````python
# run the function
sign(-45)
````

````python
# run the function
sign(0)
````

# Write a function to determine betting on a Die

Suppose I bet on a roll of a fair die. The rules of the game:<br>

- If the die shows 1 spot or 2 spots, I lose a dollar.

- If the die shows 3 spots or 4 spots, I neither lose money nor gain money.

- If the die shows 5 spots or 6 spots, I gain a dollar.

We will now use conditional statements to define a function one_bet that takes the number of spots on the roll and returns my net gain.

````python
# create an if statement to find the amount of playing the game
x = np.random.choice(np.arange(1, 7))
if x <=2:
    wins = -1
elif x <=4:
    wins = 0
elif x <=6:
    wins = 1
else:
    wins = 0

print("I won", wins)
````

````python
# create the function for playing the game
def one_bet(x):
    """earns from games"""
    if x <=2:
        return -1
    elif x <=4:
        return 0
    elif x <=6:
        return 1
    else:
        return 0
````

````python
# run function
one_bet(10)
````

````python
# run function
one_bet(13)
````

# Iteration
Repeating a process multiple times.
```
for i in make_array(element1, element2, element3, ...):
    statements
```

````python
# create the function above without passing it a value
def bet_on_one_roll():
    """roll die and return results of game"""
    x = np.random.choice(np.arange(1,7))
    if x<= 2:
        return -1
    elif x <= 4:
        return 0
    elif x<= 6:
        return 1
        
````

````python
# run the bet_on_one_roll 10 times
for i in np.arange(1, 11):
    print("bet", i, "results", bet_on_one_roll())
````

````python
# if they bet 10 times how much did they make
total = 0
for i in np.arange(1, 11):
    win = bet_on_one_roll()
    total = total + win
    print("bet", i, "winning", win, "total earned",total)
````

````python
# you decide to flip a 100 coins multiple times
# create a function to flip a coin 100 times and determine the number of heads
def flip_coin():
    """ flip coin 100 times return array"""
    return np.count_nonzero(np.random.choice(make_array("Heads", "Tails"), 100) == "Heads")
````

````python
flip_coin()
````

````python
# repeat the process 15000 times
n = 15000
heads= make_array()
for i in np.arange(n):
    new_value = flip_coin()
    heads = np.append(heads, new_value)
    
````

````python
# display the number of heads
heads
````

````python
# the number of heads
len(heads)
````

````python
# Create a Table
simulation_results = Table().with_columns(
    'Repetition', np.arange(1, n+1),
    'Number of Heads', heads)
simulation_results.show(3)
````

````python
# Histogram of number of heads
simulation_results.hist("Number of Heads", bins = np.arange(30.5, 69.6, 1))
````

````python
# create a function to roll two dice and return the results
def one_simulation():
    return sum(np.random.choice(die,2))
````

````python
# run the simulation function
one_simulation()
````

````python
# Repeat the simulation 10,000 times and create the histogram
n= 10000
moves = make_array()
for i in np.arange(n):
    new_roll = one_simulation()
    moves = np.append(moves, new_roll)

results= Table().with_columns(
    'Repetition', np.arange(1, n+1),
    'Sum of Two Rolls', moves)
results.hist('Sum of Two Rolls', bins=np.arange(1.5,12.6, 1))
````

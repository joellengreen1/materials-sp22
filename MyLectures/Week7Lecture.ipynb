# Conditional Statements
Multi-line statement that allows Python to choose among different alternatives based on the truth value of an expression.  The format of a conditional statement is as follows.
```
if x > 0:
   statements
elif x < 0:
   statements
else:
   statements

```

````python
x = 0
if x > 0:
    y = 3
elif x < 0:
    y = 4
else:
    y = 5

print('The value is', y)
````

````python
# create a function to pass a value and does the if statement above

def y_value(x):
    """detrmine the results of a number based on it sign"""
    if x > 0:
        y = 3
    elif x < 0:
        y = 4
    else:
        y = 5
    return y
````

````python
y_value(100)
````

````python
# create a function that defines the sign of a number
def sign(x):
    """determine the sign of a number"""
    if x > 0:
        return "Positive"
    elif x < 0:
        return "Negative"
    else:
        return "Neither Positive no Negative"
````

````python
sign(0)
````

````python
sign(12)
````

# Write a function to determine betting on a Die

Suppose I bet on a roll of a fair die. The rules of the game:<br>

- If the die shows 1 spot or 2 spots, I lose a dollar.

- If the die shows 3 spots or 4 spots, I neither lose money nor gain money.

- If the die shows 5 spots or 6 spots, I gain a dollar.

We will now use conditional statements to define a function one_bet that takes the number of spots on the roll and returns my net gain.

````python
# create an if statement
dice = 6
if dice <= 2:
    y = -1
elif dice<= 4:
    y = 0
else:
    y = 1
    
print("The value of y is",y)    
````

````python
# create the function
def one_bet(x):
    """ return results of 1 die rolled"""
    if x <= 2:
        return -1
    elif x <= 4:
        return 0
    elif x <= 6:
        return 1
````

````python
one_bet(3)
````

````python
one_bet(78)
````

# Iteration
Repeating a process multiple times.
```
for i in make_array(element1, element2, element3, ...):
    statements
```

````python
# create the function above without passing it a value
def bet_on_one_roll():
    """roll die and result of game"""
    x = np.random.choice(np.arange(1,7))
    if x <= 2:
        return -1
    elif x <= 4:
        return 0
    elif x <= 6:
        return 1
````

````python
# run the bet_on_one_roll 10 times
for i in np.arange(1, 11):
    print("bet",i, "results", bet_on_one_roll())
    
````

````python
# if they bet 10 times how much did they make
total = 0
for i in np.arange(1, 11):
    win = bet_on_one_roll()
    total = total + win
    print("bet",i,"winnings", win,"total earned",total)
````

````python
# you decide to flip a 100 coins multiple times
# create a function to flip a coin 100 times and determine the number of heads

def one_simulated_value():
    x = np.random.choice(['Heads', 'Tails'], 100)
    return np.count_nonzero(x == 'Heads')

````

````python
# repeat the process 15000 times
n = 15000
heads = make_array() # empty array
for i in np.arange(n):
    new_value = one_simulated_value() 
    heads = np.append(heads, new_value)
    
````

````python
heads
````

````python
len(heads)
````

````python
# Create a Table
simulation_results = Table().with_columns(
            'Repetition', np.arange(1, n + 1),
            'Number of Heads', heads)
simulation_results.show(3)
````

````python
simulation_results.hist("Number of Heads", bins = np.arange(30.5, 69.6, 1))
````
